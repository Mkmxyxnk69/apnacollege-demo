```python
import numpy as np
import pandas as pd
```

```python
df=pd.read_csv('/content/placement.csv')
```

```python
df
```

```python
#applying imputation(mean)
df['cgpa']=df['cgpa'].fillna(df['cgpa'].mean())
```

```python
df
```

```python
df['placement']=df['placement'].map({'yes':1,'no':0})
```

```python
df
```

```python
unique_branch=np.unique(df['branch'])
for i in unique_branch:
    df[i] = (df['branch'] == i).astype(int)
```

```python
df
```

```python
df=df.drop('branch',axis=1)
```
